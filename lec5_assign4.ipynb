```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Sensible global defaults so every plot looks consistent
sns.set_theme(style='whitegrid', context='notebook')
plt.rcParams['figure.figsize'] = (7, 4)
plt.rcParams['figure.dpi']     = 100
```

```python
# Bar chart of membership types — gym['abonoment_type'].value_counts()
subscription_count = gym['abonoment_type'].value_counts()
print(subscription_count)
print("INDEX: ", subscription_count.index)
print("VAL: ", subscription_count.values)

fig, ax = plt.subplots()
ax.bar(subscription_count.index, subscription_count.values, color=['pink', '#19eb4e'])
ax.set(title='subscription type count', xlabel="subscription type", ylabel='Count')
plt.show()
```

```python
#Histogram of avg_time_in_gym
fig, ax = plt.subplots()
ax.hist(gym['avg_time_in_gym'], bins=15, color='seagreen', edgecolor='white')   # bins holo koto ta slice hbe
ax.set(title='Distribution of average time in Gym',
       xlabel='Avg Time', ylabel='Frequency')
plt.show()
```

```python
#Scatter plot of Age vs visit_per_week
fig, ax = plt.subplots()
ax.scatter(gym['Age'], gym['visit_per_week'],
           alpha=0.9, s=13, color="blue")    #alpha holo color density 1 hole pure color, r s holo size of the scatter plot
ax.set(title='Age vs Visit Per Week',
       xlabel='Age', ylabel='Visit Per Week')
plt.show()
```

## **Seaborn**



```python
#Box plot — gender vs Age


# outlier dekhar jonno boxplot used
import seaborn as sns


fig, axes = plt.subplots(figsize=(9, 4))

sns.boxplot(data=gym, x='gender', y='Age')
axes.set_title('Age by Gender')


plt.tight_layout()
plt.show()
```

```python
#Count plot — days_per_week
# fig, axes = plt.subplots(1, 2, figsize=(11, 4))
fig, axes = plt.subplots(figsize=(21, 9))

sns.countplot(data=gym, x='days_per_week')
axes.set_title('Days per week in Gym')

# label of X-axis 45 degree rotated
plt.xticks(rotation=90, ha='right')      #ha = Horizontal Alignment (right/left/center)

plt.tight_layout()
plt.show()

```

```python
#Bar plot — mean visit_per_week per abonoment_type
fig, ax = plt.subplots(figsize=(9, 4))
order = gym.groupby('abonoment_type')['visit_per_week'].mean().sort_values().index

sns.barplot(data=gym, x='abonoment_type',
            y='visit_per_week', order=order, ax=ax)
ax.set_title('Average visit per week per abonoment type')
ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```
